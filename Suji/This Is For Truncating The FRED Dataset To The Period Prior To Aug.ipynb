```python
from google.colab import files
uploaded = files.upload()
input_path = list(uploaded.keys())[0]
print(input_path)
```

```python
import pandas as pd

readme = pd.read_excel(input_path, sheet_name="README", header=None)
monthly = pd.read_excel(input_path, sheet_name="Monthly", parse_dates=["observation_date"])

print(monthly.tail())
```

```python
CUTOFF = pd.Timestamp("2026-07-31")

monthly_cut = monthly[monthly["observation_date"] <= CUTOFF].copy()

print("Removed rows:")
print(monthly[monthly["observation_date"] > CUTOFF])
print("\nNew last date:", monthly_cut["observation_date"].max().date())

assert monthly_cut["observation_date"].max() <= CUTOFF
```

```python
output_path = "Q4-FRED-CPI-Midwest_thru_Jul2026.xlsx"

with pd.ExcelWriter(output_path, engine="openpyxl", datetime_format="yyyy-mm-dd") as writer:
    readme.to_excel(writer, sheet_name="README", index=False, header=False)
    monthly_cut.to_excel(writer, sheet_name="Monthly", index=False)

files.download(output_path)
```
